```python
import os
import pandas as pd
import matplotlib.pyplot as plt

# Folder containing your extracted CSV files
DATA_DIR = "./CPU_Usage_and_Throughput_without_S5"   # subdirectory with the data files

# Detect CPU and THR csv files
cpu_files = [f for f in os.listdir(DATA_DIR) if f.startswith("cpu_") and f.endswith(".csv")]
thr_files = [f for f in os.listdir(DATA_DIR) if f.startswith("throughput_") and f.endswith(".csv")]

# Sort files for consistent ordering
cpu_files.sort()
thr_files.sort()

# -----------------------------
# Plot CPU Usage Curves
# -----------------------------
plt.figure(figsize=(12, 6))
for csv_file in cpu_files:
    path = os.path.join(DATA_DIR, csv_file)
    df = pd.read_csv(path)
    # Create a cleaner label from filename
    label = csv_file.replace("cpu_", "").replace(".csv", "")
    plt.plot(df["time_s"], df["cpu_usage_ms_per_s"], label=label, linewidth=2)

plt.title("Average CPU Usage Over Time", fontsize=14, fontweight='bold')
plt.xlabel("Time (s)", fontsize=12)
plt.ylabel("CPU Usage (ms/s)", fontsize=12)
plt.grid(True, alpha=0.3)
plt.legend(fontsize=10)
plt.tight_layout()
plt.show()

# -----------------------------
# Plot Throughput Curves
# -----------------------------
plt.figure(figsize=(12, 6))
for csv_file in thr_files:
    path = os.path.join(DATA_DIR, csv_file)
    df = pd.read_csv(path)
    # Create a cleaner label from filename
    label = csv_file.replace("throughput_", "").replace(".csv", "")
    plt.plot(df["time_s"], df["throughput_Mbps"], label=label, linewidth=2)

plt.title("Average Downlink Throughput Over Time", fontsize=14, fontweight='bold')
plt.xlabel("Time (s)", fontsize=12)
plt.ylabel("Throughput (Mbps)", fontsize=12)
plt.grid(True, alpha=0.3)
plt.legend(fontsize=10)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# Read and plot UE1, UE2, UE3, UE4, UE5 CSV files from Exp1_Throughput
# Note: UE5 may not have Exp1 data, so it will be handled separately
files = {
    "UE1": "Exp1_Throughput/UE1_SST_1_SD_1.csv",
    "UE2": "Exp1_Throughput/UE2_SST_1_SD_2.csv",
    "UE3": "Exp1_Throughput/UE3_SST_1_SD_3.csv",
    "UE4": "Exp1_Throughput/UE4_SST_1_SD_4.csv",
    "UE5": None  # UE5 doesn't have Exp1 data
}

# Folder containing throughput files with S5 and CPU Control
DATA_DIR = "./CPU_Usage_and_Throughput_with_S5"

# Create the plot
plt.figure(figsize=(12, 6))

# Set the starting point for S5+CPU Control data (all UEs start here)
S5_START_X = 0  # All S5+CPU Control data starts at x=0

# First pass: Collect all x values to find the global minimum
all_x_values = []

for ue_name, filename in files.items():
    try:
        # Load Exp1_Throughput data (if available)
        df_exp1_clean = None
        if filename is not None:
            df_exp1 = pd.read_csv(filename)
            df_exp1_clean = df_exp1.dropna()
        
        # Load corresponding S5+CPU Control data
        csv_file = f"throughput_{ue_name}.csv"
        path = os.path.join(DATA_DIR, csv_file)
        df_s5 = pd.read_csv(path)
        df_s5_clean = df_s5.dropna()
        
        # S5+CPU Control data starts at S5_START_X (same for all UEs)
        x_s5_shifted = df_s5_clean["time_s"] + S5_START_X
        
        if df_exp1_clean is not None and len(df_exp1_clean) > 0:
            # Shift Exp1 data backwards so it ends just before S5_START_X
            exp1_length = df_exp1_clean["x"].max() - df_exp1_clean["x"].min() + 1
            exp1_start = S5_START_X - exp1_length  # Start Exp1 data before S5_START_X
            
            # Shift Exp1 x values to start at exp1_start
            x_exp1_min = df_exp1_clean["x"].min()
            x_exp1_shifted = df_exp1_clean["x"] - x_exp1_min + exp1_start
            
            # Collect x values to find minimum
            all_x_values.extend(x_exp1_shifted.tolist())
            all_x_values.extend(x_s5_shifted.tolist())
        else:
            # For UE5 (no Exp1 data)
            all_x_values.extend(x_s5_shifted.tolist())
    except:
        pass

# Find the global minimum x value
global_min_x = min(all_x_values) if all_x_values else 0

# Second pass: Plot with all x values shifted to start from 0 and collect metrics
metrics_data = []

for ue_name, filename in files.items():
    try:
        # Load Exp1_Throughput data (if available)
        df_exp1_clean = None
        if filename is not None:
            df_exp1 = pd.read_csv(filename)
            df_exp1_clean = df_exp1.dropna()
        
        # Load corresponding S5+CPU Control data
        csv_file = f"throughput_{ue_name}.csv"
        path = os.path.join(DATA_DIR, csv_file)
        df_s5 = pd.read_csv(path)
        df_s5_clean = df_s5.dropna()
        
        # S5+CPU Control data starts at S5_START_X (same for all UEs)
        x_s5_shifted = df_s5_clean["time_s"] + S5_START_X
        
        if df_exp1_clean is not None and len(df_exp1_clean) > 0:
            # Shift Exp1 data backwards so it ends just before S5_START_X
            exp1_length = df_exp1_clean["x"].max() - df_exp1_clean["x"].min() + 1
            exp1_start = S5_START_X - exp1_length  # Start Exp1 data before S5_START_X
            
            # Shift Exp1 x values to start at exp1_start
            x_exp1_min = df_exp1_clean["x"].min()
            x_exp1_shifted = df_exp1_clean["x"] - x_exp1_min + exp1_start
            
            # Shift all x values so minimum is 0
            x_exp1_shifted = x_exp1_shifted - global_min_x
            x_s5_shifted = x_s5_shifted - global_min_x
            
            # Concatenate x values: shifted Exp1 x values + shifted S5 x values
            x_combined = pd.concat([x_exp1_shifted, pd.Series(x_s5_shifted)], ignore_index=True)
            
            # Concatenate y values: Exp1 y values + S5 throughput values
            y_combined = pd.concat([df_exp1_clean["y"], df_s5_clean["throughput_Mbps"]], ignore_index=True)
            
            print(f"Loaded {ue_name}: {len(df_exp1_clean)} Exp1 + {len(df_s5_clean)} S5+CPU = {len(x_combined)} total data points")
        else:
            # For UE5 (no Exp1 data), shift S5 data to start from 0
            x_combined = x_s5_shifted - global_min_x
            y_combined = df_s5_clean["throughput_Mbps"]
            
            print(f"Loaded {ue_name}: {len(df_s5_clean)} S5+CPU data points (no Exp1 data)")
        
        # Save the concatenated data to CSV file
        df_final = pd.DataFrame({
            'x': x_combined,
            'y': y_combined
        })
        output_csv = f"aio/{ue_name}_final_data.csv"
        df_final.to_csv(output_csv, index=False)
        print(f"Saved {ue_name} final data to {output_csv}")
        
        # Plot as a single continuous line
        plt.plot(x_combined, y_combined, 
                label=ue_name, 
                linewidth=2)

        
    except FileNotFoundError as e:
        print(f"Warning: File not found for {ue_name}: {e}")
    except Exception as e:
        print(f"Error loading {ue_name}: {e}")


plt.title("Throughput Curves: UE1, UE2, UE3, UE4, UE5 (Exp1 + S5)", fontsize=14, fontweight='bold')
plt.xlabel("Sample Index / Time (s)", fontsize=12)
plt.ylabel("Throughput (Mbps)", fontsize=12)
plt.grid(True, alpha=0.3)
plt.legend(fontsize=10)
plt.tight_layout()
plt.show()
```
